## Histograms for continues numbers

```python
# Assuming you have already selected your numeric columns into con_df
cols = ['danceability', 'energy', 'key', 'loudness', 'mode', 'speechiness', 
        'acousticness', 'instrumentalness', 'liveness', 'valence', 'tempo', 'duration_ms']
con_df = df[cols]

```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set a global, elegant plotting style
plt.style.use('seaborn-v0_8-whitegrid')

# Define the color (you can also use a color palette if you prefer variety)
navy = 'navy'

# Make sure all selected columns are numeric
if con_df.select_dtypes(include=['number']).shape[1] == len(cols):
    num_plots = len(cols)
    num_cols = 2  # Number of plots per row
    num_rows = (num_plots + num_cols - 1) // num_cols  # Round up to cover all plots

    fig, axes = plt.subplots(num_rows, num_cols, figsize=(16, 5 * num_rows))
    axes = axes.flatten()  # Flatten to simplify indexing

    for i, col in enumerate(cols):
        con_df[col].hist(ax=axes[i], bins=35, color=navy, edgecolor='white', linewidth=1)
        axes[i].set_title(f'Distribution of {col}', fontsize=14, fontweight='bold')
        axes[i].set_xlabel(col, fontsize=12)
        axes[i].set_ylabel('Frequency', fontsize=12)
        axes[i].grid(True, linestyle='--', alpha=0.6)

    # Remove any unused subplots
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])

    plt.tight_layout()
    plt.show()

else:
    print("Not all selected columns are numeric. Please check your input.")
```

## skewness

```python
def highlight(cell_value):
    highlight = 'background-color: mediumspringgreen;'
    default = ''
    negative = 'background-color: hotpink;'
    if cell_value > 1:
        return highlight
    elif cell_value < -1:
        return negative
    #else
       # return default  
pd.DataFrame(con_df.skew(),columns=['skewness']).sort_values(by='skewness', ascending=False).style.applymap(highlight)
```

## Correlation

Identify the most significant relationships, especially with popularity

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select numeric columns
numeric_columns = df.select_dtypes(include=['float64', 'int64'])

# Compute correlation matrix
correlation_matrix = numeric_columns.corr()

# Set up the figure
plt.figure(figsize=(12, 10))  # Adjust size for better clarity

# Plot heatmap
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap='coolwarm',
    linewidths=0.5,
    linecolor='white',
    square=True,
    cbar_kws={"shrink": 0.8},
    annot_kws={"size": 8}
)

plt.xticks(rotation=45, ha='right', fontsize=9)
plt.yticks(rotation=0, fontsize=9)
plt.title("Correlation Matrix of Numeric Features", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

#### The correlation matrix highlights several interesting relationships between the variables:

- **Popularity and energy**: A moderate positive correlation, suggesting that more energetic tracks tend to be more popular.
- **Popularity and danceability**: A slightly positive correlation, indicating that more danceable tracks are often more popular.
- **Loudness and energy**: A strong positive correlation, which makes sense as more energetic tracks are generally louder.
- **Instrumentalness and popularity**: A negative correlation, meaning that instrumental tracks are generally less popular.
- **Acousticness and energy**: A negative correlation, which is expected since acoustic tracks tend to be less energetic.

## Visual representations to explore song patterns

```python
# Visualization of the distribution of popularity
plt.figure(figsize=(10, 5))
sns.histplot(df["track_popularity"], bins=30, kde=True, color="blue")
plt.title("Distribution of Track Popularity")
plt.xlabel("Popularity")
plt.ylabel("Number of Tracks")
plt.show()

# Relationship between popularity and danceability
plt.figure(figsize=(10, 5))
sns.scatterplot(x=df["danceability"], y=df["track_popularity"], alpha=0.5, color="green")
plt.title("Relationship Between Danceability and Popularity")
plt.xlabel("Danceability")
plt.ylabel("Popularity")
plt.show()

# Relationship between popularity and energy
plt.figure(figsize=(10, 5))
sns.scatterplot(x=df["energy"], y=df["track_popularity"], alpha=0.5, color="red")
plt.title("Relationship Between Energy and Popularity")
plt.xlabel("Energy")
plt.ylabel("Popularity")
plt.show()
```

### Observations from the Visualizations:

#### 1. Distribution of Popularity:
Most tracks have a popularity score between 20 and 60, with a few reaching the maximum value of 100.
The distribution is slightly left-skewed, indicating that there are more less popular songs than highly popular ones.

#### 2. Relationship Between Danceability and Popularity:
There appears to be a slight positive correlation, suggesting that tracks with higher danceability tend to be more popular.
However, there is a large spread in the data, which means that danceability is not the only factor influencing popularity.

#### 3. Relationship Between Energy and Popularity:
Popularity tends to increase with the track’s energy level, confirming the correlation observed in the heatmap.
This suggests that high-energy tracks with intense sound are more likely to attract listeners.

## Top 5 Artist by Track Popularity

```python
popular = df[df['track_popularity'] > 70]
popular
```

```python
song_counts = popular['track_artist'].value_counts()

popular = popular[popular['track_artist'].isin(song_counts[song_counts >10].index)]

avg_popular = popular.groupby('track_artist')['track_popularity'].mean()

top_artist_by_avg_popular = avg_popular.nlargest(5)
top_artist_by_song_count = song_counts.nlargest(5)
```

```python
plt.bar(top_artist_by_song_count.index, top_artist_by_song_count.values)
plt.title("Top 5 Artist by Most Song")
plt.xlabel("Artist")
plt.ylabel("Total Song")
plt.tight_layout()
plt.show()
```

```python
plt.bar(top_artist_by_avg_popular.index, top_artist_by_song_count.values)
plt.title("Top 5 Artist by Popularity Track")
plt.xlabel("Artist")
plt.ylabel("Total Song")
plt.tight_layout()
plt.show()
```
